-- most item returned


```python
most_returned_items = pd.merge(returns, order_items, on='order_id', how='inner')

most_returned_items = most_returned_items.groupby('product_name').agg({'return_id': 'count'}).reset_index().rename(columns={'return_id': 'return_count'}).sort_values(by='return_count', ascending=False)

most_returned_items
```

-- top customers with most returns

```python
top5 = pd.merge(returns, orders, on='order_id', how='inner')
most_cust_returns = pd.merge(top5, customers, on='customer_id', how='inner')

most_cust_returns = most_cust_returns.groupby(by='customer_name').size().sort_values(ascending=False).head(5)

most_cust_returns
```

-- orders that have a return date before the order date
WITH order_return_dates AS (
    SELECT
        r.return_id,
        r.order_id,
        r.return_date,
        o.order_date
    FROM returns r
    JOIN orders o ON r.order_id = o.order_id
)
SELECT *
FROM order_return_dates
WHERE return_date < order_date
